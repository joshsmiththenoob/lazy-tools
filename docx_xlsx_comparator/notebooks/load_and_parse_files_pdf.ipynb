### Read Excel file

### Compare raw value

```python
def select_records(df):
    return df.loc[
        df["作物"].eq("糙米")
        & df["年度"].isin([112, 113])
        & df["指標"].isin(["產量", "產值", "種植面積", "收穫面積"])
    ].copy()


pdf_selected = select_records(pdf_df)
excel_selected = select_records(excel_df)
```

### 對齊資料，再判斷結果

```python
KEYS = ["作物", "年度", "指標"]

result_df = pdf_selected.merge(
    excel_selected,
    on=KEYS,
    how="outer",
    suffixes=("_pdf", "_excel"),
    indicator=True,
    validate="one_to_one",
)
```

```python
result_df
```

```python
def compare_status(row):
    if row["_merge"] == "left_only":
        return "Excel 缺少資料"

    if row["_merge"] == "right_only":
        return "PDF 缺少資料"

    if row["單位_pdf"] != row["單位_excel"]:
        return "單位不同"

    pdf_value = row["數值_pdf"]
    excel_value = row["數值_excel"]

    if (
        pd.isna(pdf_value)
        or pd.isna(excel_value)
        or isinstance(pdf_value, str)
        or isinstance(excel_value, str)
    ):
        return "缺值／標記，暫不比對"

    return "相同" if pdf_value == excel_value else "數值不同"


result_df["比對結果"] = result_df.apply(compare_status, axis=1)

display(result_df.drop(columns="_merge"))
```
