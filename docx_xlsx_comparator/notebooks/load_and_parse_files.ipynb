```python
def get_common_row_keys(word_df, excel_df, key_col: str):
    word_keys = set(word_df[key_col].dropna().astype(str).str.strip())
    print(word_keys)
    excel_keys = set(excel_df[key_col].dropna().astype(str).str.strip())
    print(excel_keys)
    common_keys = list(word_keys & excel_keys)
    return common_keys
```

### Compare raw value

```python
def compare_multiple_values(word_df, excel_df, key_col, selected_rows, selected_columns):
    results = []

    for row_key in selected_rows:
        # Per raw in selected_rows -> find matching raws in both DataFrames based on key_col
        word_match = word_df[
            word_df[key_col].astype(str).str.strip() == str(row_key).strip()
        ]

        excel_match = excel_df[
            excel_df[key_col].astype(str).str.strip() == str(row_key).strip()
        ]

        for col in selected_columns:
            # Then find the value in selected_columns, if the column doesn't exist, mark as "Word/Excel 找不到列"
            
            if word_match.empty:
                word_value = None
                status = "Word 找不到列"
            else:
                word_value = word_match[col].iloc[0]
                status = "可比較"


            if excel_match.empty:
                excel_value = None
                status = "Excel 找不到列"
            else:
                excel_value = excel_match[col].iloc[0]



            if word_match.empty or excel_match.empty:
                is_equal = False
            else:
                is_equal = str(word_value).strip() == str(excel_value).strip()

            results.append({
                "主要欄位": key_col,
                "列名稱": row_key,
                "比對欄位": col,
                "Word值": word_value,
                "Excel值": excel_value,
                "是否相同": is_equal,
                "狀態": status,
            })

    return pd.DataFrame(results)
```

```python
result_df = compare_multiple_values(
    word_df=word_df,
    excel_df=excel_df,
    key_col="安全控制措施",
    selected_rows=selected_rows,
    selected_columns=selected_columns
)

result_df
```
