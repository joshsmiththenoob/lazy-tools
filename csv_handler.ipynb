```python
import pandas as pd
from pathlib import Path
from utils.time_handler import TimeHandler
```

```python
time_handler = TimeHandler()
```

```python
def load_platform(path: Path) -> pd.DataFrame:
    df = pd.read_excel(path, engine="openpyxl")
    df.columns = df.iloc[0]
    df = df.iloc[1:].reset_index(drop=True)
    df = df.rename(columns={"日期":"date","日均價格":"avg_price_platform","總交易量":"volume_platform"})
    df["date"] = time_handler.to_datetime_any(df["date"])
    df["avg_price_platform"] = pd.to_numeric(df["avg_price_platform"], errors="coerce")
    df["volume_platform"] = pd.to_numeric(df["volume_platform"], errors="coerce")
    return df.dropna(subset=["date"])

def load_market(path: Path) -> pd.DataFrame: 
        df = pd.read_excel(path, engine="odf")
        df.columns = df.iloc[1]
        df = df.iloc[2:].reset_index(drop=True)
        df = df.rename(columns={"交易日期":"date","平均價":"avg_price_market","交易量":"volume_market"})
        df["date"] = time_handler.to_datetime_any(df["date"])
        df["avg_price_market"] = pd.to_numeric(df["avg_price_market"], errors="coerce")
        df["volume_market"] = pd.to_numeric(df["volume_market"], errors="coerce")
        return df.dropna(subset=["date"])

```

```python
market_path = r"platform_market_comparator\input\109-114年交易市場-蔬菜產品交易價量走勢圖資料.ods"
platform_path = r"platform_market_comparator\input\歷史價量平台-蔬菜-批發合計.xlsx"
output_path = r"compare_price_volume_result.xlsx"
```

```python
plat = load_platform(Path(platform_path))
mkt  = load_market(Path(market_path))
```

```python
df = plat.merge(mkt, on="date", how="inner").sort_values("date")
```
